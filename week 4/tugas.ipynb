```python
train_data = dataset[["Age", "Fare"]].copy()

pos_missing_train = train_data.index[
    train_data.isna().any(axis=1)
].tolist()

train = train_data.dropna()

print(train)
```

```python
test_data = test_dataset[["Age", "Fare"]].copy()

pos_missing_test = test_data.index[
    test_data.isna().any(axis=1)
]

test = test_data.dropna()

print(f"Posisi missing values test (indeks mulai 0):\n{pos_missing_test.tolist()[:20]} ...\n")


print("\nData test setelah dropna:")
print(test)

print("\nJumlah baris yang dihapus:", len(pos_missing_test))
print("Jumlah baris tersisa:", len(test))
```

```python
train_label = dataset[["Survived"]].drop(index=pos_missing_train)

print("Label training:")
print(train_label)

print("\nJumlah label training:", len(train_label))
```

```python
test_label_dataset = pd.read_csv("titanic_testlabel.csv")

assert test_dataset["PassengerId"].equals(
    test_label_dataset["PassengerId"]
), "Urutan PassengerId pada data test dan label berbeda."

test_label = test_label_dataset[["Survived"]].drop(
    index=pos_missing_test
)

print("Label test:")
print(test_label)

print("\nJumlah label test:", len(test_label))
```

```python
min_train = train.min()
max_train = train.max()

# Pembagi 1 untuk menghindari pembagian nol jika ada fitur konstan
rentang_train = (max_train - min_train).replace(0, 1)

train_data = (train - min_train) / rentang_train

print("Minimum setiap atribut training:")
print(min_train)

print("\nMaksimum setiap atribut training:")
print(max_train)

print("\nHasil normalisasi training:")
print(train_data)
```

```python
test_data = (test - min_train) / rentang_train

print("Minimum yang digunakan (dari training):")
print(min_train)

print("\nMaksimum yang digunakan (dari training):")
print(max_train)

print("\nHasil normalisasi test:")
print(test_data)
```

```python
X_train = train_data.to_numpy()
X_test = test_data.to_numpy()
y_train = train_label["Survived"].to_numpy()

# Jarak setiap data test terhadap seluruh data training
jarak = np.sqrt(
    ((X_test[:, None, :] - X_train[None, :, :]) ** 2).sum(axis=2)
)

urutan_tetangga = np.argsort(jarak, axis=1, kind="stable")

class_result = pd.DataFrame(index=test_data.index)

for k in range(1, 16):
    indeks_tetangga = urutan_tetangga[:, :k]
    label_tetangga = y_train[indeks_tetangga]

    jumlah_suara_1 = (label_tetangga == 1).sum(axis=1)
    class_result[k] = (jumlah_suara_1 > k / 2).astype(int)

print("Hasil prediksi untuk setiap k:")
print(class_result)

print("\nKolom = nilai k, baris = indeks data test.")
```
