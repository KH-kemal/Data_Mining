# Praktikum Minggu 11 — Association Rule

## Nomor 1 — Membaca dan Menampilkan Dataset

```python
import pandas as pd

dataset = pd.read_csv("transaction.csv")

print(dataset)
```

## Nomor 2 — Mengambil data untuk negara Portugal

```python
data = dataset.loc[dataset["Country"] == "Portugal"].copy()
print(data)
```

## Nomor 3 — Membuat tabel StockCode pada setiap transaksi


```python
transaksi = pd.crosstab(data["InvoiceNo"], data["StockCode"])
transaksi = transaksi.gt(0)

print(transaksi.astype(int))
```

## Nomor 4 — Mencari association rule dengan minimum support 0,2 dan confidence 0,7

```python
%pip install mlxtend
```

```python
from mlxtend.frequent_patterns import apriori, association_rules

frequent_itemsets = apriori(
    transaksi,
    min_support=0.2,
    use_colnames=True
)

rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.7
)

print("Frequent itemsets:")
print(frequent_itemsets.to_string(index=False))

print("\nAssociation rules:")
print(rules[
    ["antecedents", "consequents", "support", "confidence", "lift"]
].to_string(index=False))
```

## Kesimpulan

Praktikum ini menerapkan association rule pada data transaksi negara
Portugal untuk mengidentifikasi pola kemunculan barang secara bersamaan.
Data diubah menjadi tabel keberadaan StockCode pada setiap InvoiceNo,
kemudian algoritma Apriori digunakan untuk mencari frequent itemsets
dengan minimum support 0,2. Aturan asosiasi dibentuk menggunakan minimum
confidence 0,7, sehingga aturan yang ditampilkan memenuhi batas frekuensi
kemunculan dan tingkat keyakinan tersebut. Nilai support, confidence,
dan lift membantu menilai hubungan antarbarang, yang dapat menjadi
bahan pertimbangan dalam penyusunan rekomendasi produk atau promosi
berdasarkan pola transaksi.
